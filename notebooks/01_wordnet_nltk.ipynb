# Prise en main de WordNet (NLTK) pour le français

WordNet regroupe les mots en **synsets** (ensembles de synonymes représentant un même concept). Grâce à l'*Open Multilingual Wordnet* (OMW), on peut interroger les synsets avec des mots **français** (`lang='fra'`). C'est la brique de base du projet : un synset sera un nœud du graphe.

Ce notebook présente les opérations utilisées dans les notebooks suivants. Il nécessite `nltk` (voir `requirements.txt`).

```python
import nltk
from nltk.corpus import wordnet as wn

nltk.download('wordnet')
nltk.download('omw-1.4')
```

## 1. Synsets d'un mot

Un mot anglais ou français renvoie la liste de ses synsets (un par sens). Un même synset peut être atteint depuis plusieurs mots.

```python
wn.synsets('car')
```

```python
wn.synsets('voiture', lang='fra')
```

Le nom d'un synset est de la forme `lemme.catégorie.numéro` (catégorie : `n` nom, `v` verbe, `a` adjectif, `r` adverbe). On peut aussi accéder directement à un synset par son nom.

```python
s = wn.synset('car.n.01')
s.name(), s.pos(), s.definition()
```

## 2. Lemmes d'un synset

`lemma_names()` donne les mots (synonymes) du synset, en anglais par défaut, et en français avec `lang='fra'`.

```python
print(s.lemma_names())
print(s.lemma_names(lang='fra'))
```

## 3. Synonymes d'un mot

```python
def get_synonyms(word, pos=None, lang='eng'):
    """Lemmes (sans doublon) de tous les synsets du mot `word`.

    `lang` est la langue du mot de départ ; les synonymes sont renvoyés dans cette même langue.
    """
    synonyms = []
    for synset in wn.synsets(word, pos=pos, lang=lang):
        synonyms.extend(synset.lemma_names(lang=lang))
    return list(dict.fromkeys(synonyms))  # doublons supprimés, ordre conservé

print(get_synonyms('car'))
print(get_synonyms('voiture', lang='fra'))
```

NLTK fournit aussi `wn.synonyms(mot)`, qui renvoie une liste de synonymes **par synset** (sans le mot lui-même). Attention : cette fonction prend un *mot*, pas un nom de synset ; `wn.synonyms('car.n.01')` renvoie donc une liste vide.

```python
print(wn.synonyms('car')[:2])
print(wn.synonyms('car.n.01'))
```

## 4. Deux synsets sont-ils synonymes ?

On considère que deux synsets sont synonymes s'ils ont la **même définition** : WordNet fusionne déjà les mots de même sens dans un seul synset, donc `car.n.01` et `automobile.n.01` désignent le même synset.

```python
def synonyme(syn1, syn2):
    """Vrai si les deux synsets ont la même définition."""
    return syn1.definition() == syn2.definition()

print(synonyme(wn.synset('car.n.01'), wn.synset('automobile.n.01')))
print(synonyme(wn.synset('car.n.01'), wn.synset('boat.n.01')))
```

## 5. Parcourir les sens d'un mot

```python
for syn in wn.synsets('table', lang='eng'):
    print(syn.name(), '-', syn.definition())
```
